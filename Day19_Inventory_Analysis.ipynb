```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
print("Dataset Shape:", df.shape)

df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df['Date'] = pd.to_datetime(df['Date'])

print(df['Date'].dtype)
```

```python
print("Total Rows:", len(df))
print("Total Columns:", len(df.columns))

df.head()
```

```python
df['Inventory Turnover'] = (
    df['Cost of Goods Sold (COGS)'] /
    df['Average Inventory']
)

df[['Category', 'Cost of Goods Sold (COGS)',
    'Average Inventory', 'Inventory Turnover']].head()
```

```python
print("Average Inventory Turnover:",
      df['Inventory Turnover'].mean())
```

```python
print("Highest Turnover:",
      df['Inventory Turnover'].max())

print("Lowest Turnover:",
      df['Inventory Turnover'].min())
```

```python
category_turnover = df.groupby('Category').agg({
    'Cost of Goods Sold (COGS)': 'sum',
    'Average Inventory': 'mean',
    'Inventory Turnover': 'mean'
}).reset_index()

category_turnover
```

```python
category_turnover = category_turnover.sort_values(
    by='Inventory Turnover',
    ascending=False
)

print(category_turnover)
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=category_turnover,
    x='Category',
    y='Inventory Turnover'
)

plt.title('Category-wise Inventory Turnover')
plt.xlabel('Product Category')
plt.ylabel('Inventory Turnover Ratio')
plt.xticks(rotation=45)
plt.show()
```

```python
slow_moving = df.sort_values(
    by='Inventory Turnover',
    ascending=True
)

slow_moving.head(10)
```

```python
fast_moving = df.sort_values(
    by='Inventory Turnover',
    ascending=False
)

fast_moving.head(10)
```

```python
slow_count = df[df['Inventory Turnover'] < 1].shape[0]

print("Slow-Moving Inventory Records:", slow_count)
```

```python
df.to_csv("Day19_Inventory_Analysis_Result.csv",
          index=False)

category_turnover.to_csv(
    "Day19_Category_Turnover.csv",
    index=False
)

print("Analysis files saved successfully!")
```
